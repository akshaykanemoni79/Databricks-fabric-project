Loading the Raw data from raw_data folder to bronze each CSV file

```python
# all the required imports 
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
from pyspark.sql import functions as F
import os
os.listdir("../raw_data")
```

Reading Customer from raw folder

```python
raw_data_csv="/Workspace/Users/akshaykanemoni79@gmail.com/Databricks-fabric-project/raw_data/customers_bad.csv"
customers_df=(
    spark.read.format("csv")
       .option("header",True)
       .option("inferSchema",True)
       .load(raw_data_csv)
)
display(customer_df)
```

```python
#quick check schema
from pyspark.sql import functions as F
customers_df.printSchema()
#quick check duplicates
df_count=customer_df.groupBy("customer_id").agg(F.count("*").alias("count")).filter(col("count")>1)
display(df_count)
#quick checks of 
df_null=customer_df.filter(col("customer_id").isNull())
display(df_null)
```

Here the Raw data of customer has been inserted in Bronze schema in catalog with name customer_raw

```python
from pyspark.sql import functions as F
df_customer_bronze=customer_df.withColumn("ingested_timestamp",F.current_timestamp())
display(df_customer_bronze)
df_customer_bronze.write.format("delta").mode("overwrite").saveAsTable("workspace.bronze.customers_raw")
```

Now we will load product csv file to data frame 

```python
raw_prd_csv="/Workspace/Users/akshaykanemoni79@gmail.com/Databricks-fabric-project/raw_data/products_bad.csv"
products_df=(
    spark.read.format("csv")
    .option("header",True)
    .option("inferSchema",True)
    .load(raw_prd_csv)
)
display(products_df)
```

```python
from pyspark.sql import functions as F

# 1. Quick check on schema
products_df.printSchema()

# 2. Check for duplicate product IDs
df_prd_duplicate = (
    products_df.groupBy("product_id")
    .agg(F.count("*").alias("count"))
    .filter(F.col("count") > 1)
)
display(df_prd_duplicate)
#3. Quick check on Nulls
df_Prd_null=products_df.filter(F.col("product_id").isNull())
display(df_Prd_null)
```

Now We will load the raw data to bronze schema in the catalog with table name as products_raw

```python
from pyspark.sql import functions as F
df_products_raw=products_df.withColumn("ingested_timestamp",F.current_timestamp())
display(df_products_raw)
df_products_raw.write.format("delta").mode("overwrite").saveAsTable("workspace.bronze.products_raw")
```
